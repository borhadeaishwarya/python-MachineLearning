```python
df=pd.read_excel(r"C:\Users\DELL\Desktop\Aish\advanced-filter.xlsx",header=3)
df=df.dropna(axis=1,how='all')
df.head(4)
```

```python
df=df[['Date','Sales Rep','Region','Sales']]
df
```

```python
2. Histogram

# Use a histogram when you want to understand the distribution of numerical data.
```

```python
plt.hist(df["Region"],bins=50
        )
plt.title("Extract unique list")
plt.xlabel("Region")
plt.ylabel("Sales")
plt.show()
```

```python
plt.hist(df['Sales Rep'],bins=20)
plt.title('Extract unique list')
plt.xlabel('Sales Rep')
plt.ylabel('Sales')
plt.show()
```

```python
plt.hist(df['Sales'], bins=5, edgecolor='black')

plt.title('Distribution of Sales')
plt.xlabel('Sales')
plt.ylabel('Frequency')

plt.show()
```

```python
# 3. Box Plot

# A box plot is useful for understanding: Minimum,Maximum,Median,Spread,Outliers
```

```python
plt.boxplot(df['Sales'])

plt.title('Sales Distribution')
plt.ylabel('Sales')

plt.show()
```

```python
sns.boxplot(x=df['Sales'])

plt.title('Sales Distribution')

plt.show()
```

```python
sns.boxplot(data=df, x='Region', y='Sales')

plt.title('Sales Distribution by Region')
plt.xlabel('Region')
plt.ylabel('Sales')

plt.show()
```

```python
# 4. Scatter Plot
# Use a scatter plot when you want to see the relationship between two numerical
```

```python
df['Date'] = pd.to_datetime(df['Date'], format='%m-%d-%Y')
```

```python
df['Day'] = df['Date'].dt.day
```

```python
plt.scatter(df['Day'], df['Sales'])

plt.title('Sales vs Day')
plt.xlabel('Day')
plt.ylabel('Sales')

plt.show()
```

```python
# hue means:

# Use different colors for different categories.

# For your dataset, we can use Region as the hue.
```

```python
sns.scatterplot(data=df, x='Day', y='Sales',  hue='Region')

plt.title('Sales vs Day')

plt.show()
```

```python
df['Day'] = df['Date'].dt.day
df['Month'] = df['Date'].dt.month
df['Year'] = df['Date'].dt.year
```

```python
sns.scatterplot(data=df, x='Day', y='Sales')
plt.xticks(rotation=45)
plt.title('Sales vs Day')

plt.show()
```

```python
#5. Bar Chart
# Use a bar chart when you want to compare categories.
```

```python
region_sales = df.groupby('Region')['Sales'].sum()

print(region_sales)
```

```python
plt.bar(region_sales.index, region_sales.values)

plt.title('Total Sales by Region')
plt.xlabel('Region')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.show()
```

```python
sns.barplot(data=df, x='Region', y='Sales', estimator='sum')

plt.title('Total Sales by Region')

plt.show()
```

```python
# 6. Count Plot

# This one is slightly different from a bar chart.

# Use a count plot when you want to count how many times each category occurs.
# Count plot counts rows.
# It does not calculate the total sales amount.
#  ex -How many sales records does each Sales Rep have?
```

```python
sns.countplot(data=df, x='Sales Rep')

plt.title('Number of Sales by Sales Rep')
plt.xlabel('Sales Rep')
plt.ylabel('Count')

plt.show()
```

```python
sns.barplot(
    data=df,
    x='Sales Rep',
    y='Sales',
    estimator='sum'
)
```
